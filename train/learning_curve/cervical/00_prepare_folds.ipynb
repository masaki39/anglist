# 頚椎学習曲線 — fold 準備

omuro データセット（324件）を 5-fold CV に分割し、`cervical_folds.json` を生成する。

**1回のみ実行**。以降は `01_train.ipynb` を使う。

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# ===== 設定 =====
DRIVE_DATA_DIR = '/content/drive/MyDrive/spine_data/omuro_cervical'
DRIVE_LC_DIR   = '/content/drive/MyDrive/spine_data/omuro_cervical_lc'
N_FOLDS = 5
SEED    = 42
# ================
```

```python
import json, os, random

cases = sorted(
    f.replace('_image.npy', '')
    for f in os.listdir(DRIVE_DATA_DIR)
    if f.endswith('_image.npy')
)
print(f'Total cases: {len(cases)}')

random.seed(SEED)
shuffled = cases[:]
random.shuffle(shuffled)

folds = {}
for k in range(N_FOLDS):
    folds[str(k + 1)] = shuffled[k::N_FOLDS]

for k, v in folds.items():
    print(f'  fold {k}: {len(v)} cases')

os.makedirs(DRIVE_LC_DIR, exist_ok=True)
out = os.path.join(DRIVE_LC_DIR, 'cervical_folds.json')
with open(out, 'w') as f:
    json.dump({'n_folds': N_FOLDS, 'seed': SEED, 'folds': folds}, f, indent=2)
print(f'Saved: {out}')
```
